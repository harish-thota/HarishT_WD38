```python
dbutils.widgets.dropdown("DE dropdown","Azure Data Engineer",["Azure Data Engineer","Snowflake Data Engineer","Databricks Data Engineer"],"dropdown of DE roles")
dbutils.widgets.multiselect("DE Multi-Select","Azure Data Engineer",["Azure Data Engineer","Snowflake Data Engineer","Databricks Data Engineer"],"Multi-Select of DE roles")
dbutils.widgets.combobox("DE Combo-Box","Azure Data Engineer",["Azure Data Engineer","Snowflake Data Engineer","Databricks Data Engineer"],"Combo-Box of DE roles")
```
